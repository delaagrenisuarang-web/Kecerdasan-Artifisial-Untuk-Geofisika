##### **5 Prompt Geofisika dengan Struktur R–T–C–I–K–O–Q–V**
##### *Struktur Prompt*
- **R (Role)** &rarr; Peran AI (Menentukan identitas, kepribadian, latar belakang keahlian, atau posisi spesifik yang harus diadopsi oleh AI).
- **T (Task)** &rarr; Instruksi atau tindakan spesifik yang harus diselesaikan oleh AI.
- **C (Context)** &rarr; Informasi latar belakang, situasi, alasan, atau lingkungan tempat masalah tersebut terjadi.
- **I (Input)** &rarr; Informasi, angka, teks, berkas, atau parameter mentah yang disediakan pengguna untuk diproses oleh AI.
- **K (Constraints)** &rarr; Batasan atau ketentuan yang harus perhatikan oleh AI.
- **O (Output)** &rarr; Format, bentuk, atau susunan luaran akhir yang diharapkan pengguna.
- **Q (Quality Criteria)** &rarr; Standar kelayakan, kualifikasi, atau indikator keberhasilan dari hasil analisis AI.
- **V (Verification)** &rarr; Instruksi bagi AI untuk melakukan validasi silang, pengujian logika, atau pemeriksaan ulang sebelum memberikan jawaban akhir.

##### **PROMPT 1 : PEMROSESAN DATA MIKROSEISMIK (HVSR)**
*Sebagai ahli seismologi terapan dan dosen pembimbing praktikum Geofisika Lingkungan, lakukan analisis kurva HVSR (H/V Spectrum Ratio) menggunakan data sinyal mikroseismik 3-komponen (Z, N, E) durasi 30 menit dan $v_s = 200\text{ m/s}$ untuk mengidentifikasi frekuensi dominan ($f_0$), faktor amplifikasi ($A_0$), indeks kerentanan seismik ($K_g = A_0^2 / f_0$), serta ketebalan sedimen ($h = v_s / (4 \times f_0)$) guna pemetaan risiko bencana di dekat zona sesar aktif. Analisis wajib menerapkan bandpass filter (0.5–20 Hz), smoothing Konno-Ohmachi, dan disajikan tanpa sudut pandang orang pertama. Luaran laporan mencakup penjelasan alur pemrosesan, nilai parameter hasil perhitungan, kode Python (ObsPy/SciPy) lengkap dengan grafik kurva H/V, tabel ringkasan, serta dua paragraf interpretasi zonasi kerentanan yang terverifikasi sesuai standar SESAME (2004).*

##### **R — Role**
Berperan sebagai ahli seismologi terapan dan dosen pembimbing praktikum Geofisika Lingkungan.

##### **T — Task**
Lakukan analisis kurva HVSR (H/V Spectrum Ratio) untuk mengidentifikasi frekuensi dominan ($f_0$) dan faktor amplifikasi ($A_0$), lalu hitung indeks kerentanan seismik ($K_g$) serta estimasi ketebalan lapisan sedimen.

##### **C — Context**
 Analisis dilakukan untuk pemetaan tingkat kerentanan tanah di wilayah permukiman dekat zona sesar aktif guna mitigasi risiko bencana     gempa bumi. Hasil analisis akan dimasukkan dalam laporan praktikum mahasiswa S1 Geofisika.

##### **I — Input**
- Data sinyal mikroseismik 3-komponen (Z, N, E) format ASCII/CSV durasi 30 menit.
- Kecepatan gelombang geser rata-rata lapisan sedimen ($v_s$) sebesar 200 m/s.

##### **K — Constraints**
- Terapkan bandpass filter (0.5–20 Hz) dan smoothing Konno-Ohmachi.
- Tampilkan seluruh tahapan rumus dan perhitungan secara runut.
- Gunakan persamaan indeks kerentanan $K_g = A_0^2 / f_0$ dan estimasi ketebalan sedimen $h = v_s / (4 \times f_0)$.

##### **O — Output**
Berikan:
1. Penjelasan langkah pemrosesan data HVSR.
2. Nilai frekuensi dominan ($f_0$) dan amplifikasi ($A_0$).
3. Perhitungan rinci nilai $K_g$ dan $h$.
4. Kode Python menggunakan modul ObsPy/SciPy untuk pengolahan sinyal dan plotting kurva H/V.
5. Tabel ringkasan parameter HVSR.
6. Interpretasi zonasi kerentanan dalam dua paragraf.

##### **Q — Quality Criteria**
Perhitungan tepat, kode Python dapat dijalankan tanpa error, dan format laporan dapat langsung ditulis ulang oleh mahasiswa.

##### **V — Verification**
Pastikan satuan frekuensi (Hz), kecepatan (m/s), dan ketebalan (m) sudah konsisten serta sesuai dengan kriteria puncak spektrum SESAME (2004).

##### **PROMPT 2 : PEMODELAN INVERSI GEOLISTRIK TAHANAN JENIS 2D**
*Sebagai ahli geofisika eksplorasi dan instruktur praktikum Geolistrik & Elektroprospeksi, lakukan pemodelan maju (forward modeling) dan inversi data resistivitas semu 1D/2D konfigurasi Schlumberger ($AB/2$ dan $MN/2$) menggunakan pendekatan damped least-squares (algoritma Marquardt-Levenberg) untuk mengidentifikasi kedalaman, ketebalan, dan keberadaan lapisan akuifer air tanah tertekan di daerah terdampak kekeringan. Analisis ditulis tanpa sudut pandang orang pertama dengan menyajikan penurunan matematis fungsi objektif, nilai resistivitas sejati ($\rho$), ketebalan ($h$), kedalaman ($d$), nilai RMSE, kode Python (NumPy/SciPy) untuk kalkulasi dan plot kurva matching, tabel parameter lapisan, serta dua paragraf interpretasi hidrogeologi bawah permukaan dengan target RMSE di bawah 5% dan iterasi model yang stabil.*

##### **R — Role**
Berperan sebagai ahli geofisika eksplorasi dan instruktur praktikum Geolistrik & Elektroprospeksi.

##### **T — Task**
Lakukan pemodelan maju (*forward modeling*) dan inversi data resistivitas semu 1D/2D konfigurasi Schlumberger untuk mendeteksi keberadaan, kedalaman, serta ketebalan lapisan akuifer air tanah tertekan.

##### **C — Context**
Pengolahan data dilakukan untuk membantu perencanaan pembuatan sumur bor di daerah terdampak kekeringan. Analisis ini menjadi studi kasus dalam tugas praktikum pemrograman geofisika mahasiswa S1 Geofisika.

##### **I — Input**
- Data lapangan berupa tabel spasi elektroda arus ($AB/2$) dan elektroda potensial ($MN/2$).
- Nilai resistivitas semu observasi ($\rho_a$) dalam satuan $\Omega\text{m}$.

##### **K — Constraints**
- Terapkan pendekatan inversi *damped least-squares* (algoritma Marquardt-Levenberg).
- Tampilkan seluruh tahapan formulasi matematis secara runut.
- Gunakan nilai batas resistivitas batuan standar untuk interpretasi litologi dan akuifer.

##### **O — Output**
Berikan:
1. Penurunan matematis ringkas fungsi objektif pemodelan maju dan inversi.
2. Nilai resistivitas sejati ($\rho$), ketebalan ($h$), dan kedalaman ($d$) hasil inversi.
3. Perhitungan rinci nilai kesalahan acak (*misfit*) atau RMSE.
4. Kode Python menggunakan modul NumPy/SciPy untuk kalkulasi pemodelan inversi dan plotting kurva *matching* (data observasi vs kalkulasi).
5. Tabel ringkasan parameter lapisan geolistrik.
6. Interpretasi hidrogeologi bawah permukaan dalam dua paragraf.

##### **Q — Quality Criteria**
Perhitungan tepat dengan nilai RMSE di bawah 5%, kode Python dapat dijalankan tanpa error, dan format laporan dapat langsung ditulis ulang oleh mahasiswa.

##### **V — Verification**
Pastikan stabilitas iterasi inversi terverifikasi untuk mencegah terjadinya *overfitting* atau *underfitting* pada model lapisan akhir.

##### **PROMPT 3 : PEMISAHAN ANOMALI REGIONAL-SISA DATA GAYA BERAT**

Tanpa menggunakan sudut pandang orang pertama, sajikan penjelasan dasar teori frekuensi, perhitungan determinasi matriks, nilai koefisien regresi, kode Python (NumPy/SciPy/Matplotlib) untuk kalkulasi serta visualisasi peta kontur 2D (ABL, Regional, Residual), tabel statistik, dan dua paragraf interpretasi struktur geologi lokal berdasarkan evaluasi residual sum of squares.*

##### **R — Role**
Berperan sebagai ahli geofisika komputasi dan dosen pembimbing tugas akhir bidang Metode Gaya Berat (Gravity).

##### **T — Task**
Pisahkan anomali gaya berat regional dan sisa (*residual*) dari data Anomali Bouguer Lengkap (ABL) menggunakan teknik *Polynomial Trend Surface Analysis*.

##### **C — Context**
Pemisahan anomali dilakukan untuk mengisolasi respons struktur dangkal (seperti zona patahan atau reservoir panas bumi) dari pengaruh batuan dasar (*basement*) yang dalam. Hasil analisis akan dimasukkan dalam laporan praktikum mahasiswa S1 Geofisika.

##### **I — Input**
- Dataset spasial koordinat UTM ($X, Y$) dalam format CSV.
- Nilai Anomali Bouguer Lengkap ($ABL$) dalam satuan mGal.

##### **K — Constraints**
- Terapkan pemodelan regresi polinomial permukaan orde-1 dan orde-2.
- Tampilkan seluruh tahapan penyusunan matriks dan rumus regresi secara runut.
- Gunakan format LaTeX untuk seluruh notasi dan persamaan matematika.

##### **O — Output**
Berikan:
1. Penjelasan dasar teori pemisahan anomali frekuensi tinggi dan frekuensi rendah.
2. Nilai koefisien regresi polinomial dan nilai varians sisa.
3. Perhitungan rinci determinasi matriks dan pemisahan nilai anomali.
4. Kode Python menggunakan modul NumPy/SciPy/Matplotlib untuk kalkulasi matriks dan plotting peta kontur 2D (ABL, Regional, dan Residual).
5. Tabel ringkasan koefisien dan statistik pemisahan anomali.
6. Interpretasi struktur geologi berdasarkan peta anomali residual dalam dua paragraf.

##### **Q — Quality Criteria**
Perhitungan tepat, peta kontur residual menampilkan pola anomali lokal terisolasi secara jelas, kode Python dapat dijalankan tanpa error, dan format laporan dapat langsung ditulis ulang oleh mahasiswa.

Pastikan nilai *residual sum of squares* antara polinomial orde-1 dan orde-2 telah diverifikasi untuk memilih orde terbaik yang paling mewakili tren regional.

##### **PROMPT 4 : ANALISIS LINGKARAN MOHR DAN GEOMEKANIK BATUAN**
*Sebagai ahli geomekanika batuan dan dosen pengampu mata kuliah Geofisika Reservoir, hitung tegangan-tegangan utama ($\sigma_1, \sigma_3$), tegangan geser maksimum ($\tau_{max}$), dan plot Lingkaran Mohr berdasarkan data tegangan normal ($\sigma_x = 80\text{ MPa}$, $\sigma_y = 30\text{ MPa}$), tegangan geser ($\tau_{xy} = 25\text{ MPa}$), kohesi ($C = 10\text{ MPa}$), dan sudut geser dalam ($\phi = 30^\circ$) untuk mengevaluasi potensi kegagalan geser batuan reservoir panas bumi akibat tekanan fluida. Analisis dilakukan tanpa sudut pandang orang pertama dengan menerapkan kriteria Mohr-Coulomb, menyajikan perhitungan sudut bidang kritis ($\theta$), kode Python (Matplotlib/NumPy) berasio aspek 1:1, tabel parameter geomekanika, serta dua paragraf interpretasi tingkat keamanan reservoir yang terverifikasi melalui nilai eigen matriks tensor tegangan.*

##### **R — Role**
Berperan sebagai ahli geomekanika batuan dan dosen pengampu mata kuliah Geofisika Reservoir.

##### **T — Task**
Hitung tegangan-tegangan utama ($\sigma_1, \sigma_3$), tegangan geser maksimum ($\tau_{max}$), dan plot Lingkaran Mohr untuk mengevaluasi potensi kegagalan geser (*shear failure*) pada batuan reservoir akibat tekanan fluida.

##### **C — Context**
Evaluasi dilakukan untuk menilai batas aman tekanan injeksi fluida pada sumur panas bumi agar tidak memicu reaktivasi sesar lokal atau *induced seismicity*. Hasil analisis akan dimasukkan dalam laporan praktikum mahasiswa S1 Geofisika.

##### **I — Input**
- Tegangan normal sumbu-$x$ ($\sigma_x$) = $80\text{ MPa}$ dan sumbu-$y$ ($\sigma_y$) = $30\text{ MPa}$.
- Tegangan geser ($\tau_{xy}$) = $25\text{ MPa}$.
- Parameter Kriteria Kegagalan: Kohesi ($C$) = $10\text{ MPa}$ dan Sudut geser dalam ($\phi$) = $30^\circ$.

##### **K — Constraints**
- Terapkan kriteria kegagalan Mohr-Coulomb ($\tau = C + \sigma \tan\phi$).
- Tampilkan seluruh tahapan rumus dan perhitungan secara runut.
- Gunakan format LaTeX untuk seluruh variabel dan persamaan matematika.

##### **O — Output**
Berikan:
1. Penjelasan konsep keadaan tegangan dan kriteria kegagalan batuan.
2. Nilai pusat lingkaran ($C_{mohr}$), jari-jari ($R$), $\sigma_1$, $\sigma_3$, dan $\tau_{max}$.
3. Perhitungan rinci sudut orientasi bidang kritis kegagalan ($\theta$).
4. Kode Python menggunakan modul Matplotlib/NumPy untuk kalkulasi tensor tegangan dan plotting Lingkaran Mohr beserta garis batas Mohr-Coulomb.
5. Tabel ringkasan parameter tegangan utama dan geomekanika.
6. Interpretasi tingkat keamanan batuan reservoir dalam dua paragraf.

##### **Q — Quality Criteria**
Perhitungan tepat, plot grafik memiliki *aspect ratio* seimbang (1:1), kode Python dapat dijalankan tanpa error, dan format laporan dapat langsung ditulis ulang oleh mahasiswa.

##### **V — Verification**
Pastikan nilai tegangan utama ($\sigma_1, \sigma_3$) sudah terverifikasi menggunakan metode perhitungan nilai eigen (*eigenvalue*) dari matriks tensor tegangan $2 \times 2$.

Seluruh tahapan ditulis tanpa sudut pandang orang pertama, menyajikan alur matematis lengkap, nilai rata-rata petrofisika zona target, kode Python (Pandas/Matplotlib) untuk pemrosesan data dan visualisasi log plot 3-track vertikal, tabel ringkasan interval kedalaman, serta dua paragraf interpretasi prospek reservoir dengan batasan nilai $S_w \le 100\%$ dan porositas realistis (0–35%).*

Berperan sebagai ahli petrofisika dan dosen pengampu praktikum Evaluasi Formasi.

##### **T — Task**

Analisis bertujuan untuk menentukan ketebalan bersih (*net pay zone*) dari formasi batupasir selang-seling serpih (*shaly-sand*) yang berpotensi menjadi reservoir hidrokarbon atau akuifer air bersih. Hasil analisis akan dimasukkan dalam laporan praktikum mahasiswa S1 Geofisika.
